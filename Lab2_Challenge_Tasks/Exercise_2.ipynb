```python
# [CELL 1] CONFIGURATION 
ENGINEER = "Quinsayas"
STUDENT_ID ="TUPM-26-1928"
SEED_NUM = int(STUDENT_ID[-1])
print(f"System Initialized by: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] GENERATE SIGNAL
SIGNAL_LENGTH = 20

value = sum(ord(ch) for ch in ENGINEER) * SEED_NUM
signal = ""

for i in range(SIGNAL_LENGTH):
    value = (value * 31 + SEED_NUM + i) % 1000
    signal += chr(32 + value % 95)

print(f"Generated Signal: '{signal}'")
```

```python
# [CELL 3] PROCESS CHARACTER BY CHARACTER
letters = digits = spaces = others = 0
processed = ""
log = []

for pos, ch in enumerate (signal, start=1):
    if ch.isalpha():
        letters += 1
        processed += ch.upper()
        kind = "LETTER"
    elif ch.isdigit():
        digits += 1
        processed += ch
        kind = "DIGIT"
    elif ch =="":
        spaces += 1
        processed += "_"
        kind = "SPACE"
    else:
        others += 1
        processed += "?"
        kind = "OTHER"
    log.append(f"Char {pos}/{SIGNAL_LENGTH}:'(ch)' (ASCII {ord(ch)} -> kind)")
```

```python
# [CELL 4] SIGNAL CLASSIFICATION
valid = letters + digits
percent = (valid / SIGNAL_LENGTH) * 100

if percent >= 80:
    classification = "STRONG SIGNAL"
elif percent >= 50:
    classification = "WEAK SIGNAL"
else:
    classification = "CORRUPTED SIGNAL"
```

```python
# [CELL 5] EXECUTION LOG
print("---EXECUITON LOG---")
for line in log:
    print(line)
```

```python
# [CELL 6] FINAL OUTPUT
print("---DIAGNOSTIC REPORT---")
print(f"Engineer: {ENGINEER} | Seed: {SEED_NUM}")
print(f"Generated Signal: '{signal}'")
print(f"Processed Signal: '{processed}'")
print(f"Letters: {letters} | Digits: {digits} | Spaces: {spaces} | Others: {others}")
print(f"Valid Characters: {valid} / {SIGNAL_LENGTH} ({percent:.1f}%)")
print(f"Classification: {classification}")
```
