```python
# [CELL 1] CONFIGURATION 
ENGINEER = "Quinsayas"
STUDENT_ID ="TUPM-26-1928"
SEED_NUM = int(STUDENT_ID[-1])
print(f"System Initialized by: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] GENERATE CREDENTIAL AND LIMIT
code = (sum(ord(ch) for ch in ENGINEER) * SEED_NUM) % 1000
PASSWORD = ENGINEER[:3].upper() + str(code).zfill(3)
ATTEMPT_LIMIT = (SEED_NUM % 3) + 3

print(f"Generated Password: {PASSWORD}")
print(f"Attempt Limit: {ATTEMPT_LIMIT}")
```

```python
# [CELL 3] AUTHENTICATION
attempts = 0
access = "DENIED"
state = "LOCKED"
log = []

while attempts < ATTEMPT_LIMIT:
    guess = input(f"Enter Password (Attempt {attempts + 1}/{ATTEMPT_LIMIT}:").strip()
    attempts += 1
    if guess == PASSWORD:
        log.append(f"Attempt {attempts}: '{guess}' -> CORRECT")
        access = "GRANTED"
        state = "UNLOCKED"
        break
    else:
        left = ATTEMPT_LIMIT - attempts
        log.append(f" Attempts {attempts}: '{guess}' -> WRONG ({left} left)")
        print(f"Access Denied. {left} attempt(s) left.")

if access == "DENIED":
    state = "LOCKED OUT (AUTHENTICATION TERMINATED)"
```

```python
# [CELL 4] EXECUTION LOG
print("---EXECUTION LOG---")
for line in log:
    print(line)

print("---FINAL OUTPUT---")
print(f"Engineer: {ENGINEER} | Seed Value: {SEED_NUM}")
print(f"Generated Password: {PASSWORD}")
print(f"Attempt Limit: {ATTEMPT_LIMIT}")
print(f"Attempts Made: {attempts}")
print(f"Access Result: {access}")
print(f"Final System State: {state}")
```
