```python
# [CELL 1] CONFIGURATION 
ENGINEER = "Quinsayas"
STUDENT_ID ="TUPM-26-1928"
SEED_NUM = int(STUDENT_ID[-1])
print(f"System Initialized by: {ENGINEER}")
print(f"Algorithmic Seed Value: {SEED_NUM}")
```

```python
# [CELL 2] GENERATIVE SENSOR DATA
NUM_READINGS = 10

value = sum(ord(ch) for ch in ENGINEER) * SEED_NUM
readings = []
print(f"---GENERATING {NUM_READINGS} SENSOR READINGS---")
for i in range (NUM_READINGS):
    value = (value * 31 + SEED_NUM + i) % 100
    readings.append(value)
    print(f"Reading {i + 1}: {value}")
print(f"Generated Sensor Data: {readings}")
```
